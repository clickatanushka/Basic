## 1. Variables are labelled boxes
`name = value` puts a value in a box with a label. We use order **O10006** (2 x Denim Jeans, Delhi, UPI).

```python
order_id = "O10006"     # text  -> str
city = "Delhi"          # text  -> str
quantity = 2            # whole number -> int
unit_price = 1440       # whole number -> int
discount = 0.05         # decimal -> float
is_returned = False     # True/False -> bool
print(order_id, city, quantity, unit_price, discount, is_returned)
```

## 2. Checking types with `type()`

```python
print(type(order_id))
print(type(quantity))
print(type(discount))
print(type(is_returned))
```

## 3. Calculating revenue
Revenue = quantity x unit price x (1 - discount). Same formula as the Excel column in Module 2.

```python
revenue = quantity * unit_price * (1 - discount)
print(revenue)
print(type(revenue))   # int x int x float -> float
```

```python
# round(value, 2) keeps 2 decimal places, like money
revenue = round(revenue, 2)
revenue
```

## 4. f-strings: put variables inside a sentence

```python
print(f"Order {order_id} from {city}: revenue is ₹{revenue}")
```

## 5. Text that looks like a number is NOT a number
In `orders_raw.csv`, 153 prices are stored as text like `"₹1,299"`. Python cannot do maths on them until we clean them.

```python
messy_price = "₹1,299"
print(type(messy_price))

clean_text = messy_price.replace("₹", "").replace(",", "")   # remove symbol and comma
print(clean_text, type(clean_text))

price = int(clean_text)    # convert text -> whole number
print(price, type(price))
print(price * 2)           # now maths works
```

Other conversions: `float("0.05")` gives 0.05, `str(1440)` gives "1440", `int(2.9)` gives 2 (it chops, it does not round).

```python
print(float("0.05"), str(1440), int(2.9))
```

## 6. Useful string tricks for messy city names

```python
raw_city = " mumbai "
print(raw_city.strip())          # remove spaces at both ends
print(raw_city.strip().title())  # Mumbai
print("DELHI".title())           # Delhi
print("delhi" == "Delhi")        # False: Python is case-sensitive
```
