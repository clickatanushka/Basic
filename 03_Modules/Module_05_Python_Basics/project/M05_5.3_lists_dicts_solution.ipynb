## 1. Lists: many values in order
Revenue of the first six ShopKart orders of 2025 (O10001 to O10006).

```python
order_amounts = [1737.6, 570.0, 2145.1, 3434.0, 3486.6, 2736.0]
print(order_amounts)
print(len(order_amounts))    # how many
print(sum(order_amounts))    # total
print(max(order_amounts), min(order_amounts))
```

```python
# Average order value = total / count
round(sum(order_amounts) / len(order_amounts), 2)
```

## 2. Indexing and slicing (Python counts from 0)
| position | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|
| value | 1737.6 | 570.0 | 2145.1 | 3434.0 | 3486.6 | 2736.0 |

```python
print(order_amounts[0])     # first
print(order_amounts[-1])    # last
print(order_amounts[0:3])   # positions 0,1,2 (stop is NOT included)
```

```python
cities = ["Delhi", "Chennai", "Delhi", "Lucknow"]
cities.append("Chennai")      # add to the end
print(cities)
print(cities.count("Delhi"))  # how many times Delhi appears
print("Pune" in cities)       # membership check
```

```python
# Lists can be changed: replace the item at position 3 (the 4th item)
cities[3] = "Pune"
print(cities)
sorted(order_amounts, reverse=True)   # biggest first (new list, original unchanged)
```

## 3. Dictionaries: label -> value
A dictionary describes ONE thing with named fields. This is customer **C0593** from `customers.csv`.

```python
customer = {
    "customer_id": "C0593",
    "name": "Kavya Khan",
    "gender": "Female",
    "age": 26,
    "city": "Delhi",
    "membership": "Prime",
}
print(customer["name"])
print(customer["membership"])
```

```python
customer["age"] = 27                    # update a value (birthday!)
customer["orders_2025"] = 7             # add a new key
print(customer.keys())
print(customer.get("phone", "not stored"))   # safe lookup: no error if missing
```

## 4. A list of dictionaries is a mini table
Each dictionary = one row. Each key = one column name. This is exactly how a spreadsheet or SQL table looks.

```python
orders = [
    {"order_id": "O10001", "city": "Delhi",   "product": "Wings of Fire",   "revenue": 1737.6, "returned": "N"},
    {"order_id": "O10002", "city": "Chennai", "product": "Face Wash",       "revenue": 570.0,  "returned": "N"},
    {"order_id": "O10003", "city": "Delhi",   "product": "Saree",           "revenue": 2145.1, "returned": "N"},
    {"order_id": "O10004", "city": "Lucknow", "product": "Pressure Cooker", "revenue": 3434.0, "returned": "Y"},
]
print(len(orders), "rows")
print(orders[3]["product"], orders[3]["returned"])   # row 3, column product / returned
```

```python
# Sneak peek of Module 6: pandas turns this list of dictionaries into a real table
import pandas as pd
pd.DataFrame(orders)
```
