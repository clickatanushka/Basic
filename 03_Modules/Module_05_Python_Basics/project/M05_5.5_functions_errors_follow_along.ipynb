## 2. Reading error messages
Errors are normal. Read the **last line first**: it says the error type and the reason. Then look up for the line number / arrow.

### IndentationError
```python
amount = 1299
if amount > 1000:
print("Big order")
```
Output:
```
IndentationError: expected an indented block after 'if' statement on line 2
```
Fix: indent the line after the `if` by 4 spaces.

### NameError
```python
total = 5
print(totl)
```
Output:
```
NameError: name 'totl' is not defined. Did you mean: 'total'?
```
Fix: a typo, or you forgot to run the cell that creates the variable (common after a runtime restart).

### TypeError: adding text and a number
This is exactly what happens with the `"₹1,299"` prices in `orders_raw.csv`.

The cells below use `try / except` only so the notebook keeps running and prints the real message. You do not need to write `try` yourself yet.

````python
price = "₹1,299"     # text, as stored in the raw file
try:
    print(price + 100)
except Exception as e:
    print(type(e).__name__, ":", e)
````

````python
# ValueError: the type is OK (text) but the content cannot become a number
try:
    int("₹1,299")
except Exception as e:
    print(type(e).__name__, ":", e)
````

````python
# KeyError: asking a dictionary for a key it does not have
customer = {"customer_id": "C0593", "name": "Kavya Khan", "city": "Delhi"}
try:
    customer["phone"]
except Exception as e:
    print(type(e).__name__, ":", e)
````

| Error | Plain meaning | Usual fix |
|---|---|---|
| IndentationError | Spaces at line start are wrong | Indent 4 spaces after `if`, `for`, `def` |
| NameError | Python never saw that name | Fix typo / run the earlier cell |
| TypeError | Wrong kind of value for this operation | Convert with `int()`, `float()`, `str()` |
| ValueError | Right kind, bad content | Clean the text first (remove ₹ and commas) |
| KeyError | Key not in dictionary / column not in table | Check spelling and capitals |
| FileNotFoundError | File not in Colab Files panel | Upload it again |
