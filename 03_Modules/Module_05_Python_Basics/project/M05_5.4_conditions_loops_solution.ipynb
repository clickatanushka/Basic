## 1. if / elif / else
ShopKart rule: revenue of ₹5,000 or more is a **Big** order, ₹1,000 or more is **Medium**, else **Small**.
Note the colon `:` and the 4-space indent - Python uses indentation instead of brackets.

```python
revenue = 3434.0
if revenue >= 5000:
    size = "Big"
elif revenue >= 1000:
    size = "Medium"
else:
    size = "Small"
print(size)
```

Comparison operators: `==` equal, `!=` not equal, `>`, `<`, `>=`, `<=`. Combine with `and`, `or`, `not`.

```python
city = "Pune"
payment_mode = "COD"
if city == "Pune" and payment_mode == "COD":
    print("Flag for return-risk check")
```

## 2. for loops: do the same thing for every item

```python
order_amounts = [1737.6, 570.0, 2145.1, 3434.0, 3486.6, 2736.0, 24484.8, 14098.95]
for amount in order_amounts:
    print(amount)
```

```python
# Running total and counting big orders - what SUM and COUNTIF do in Excel
total = 0
big_orders = 0
for amount in order_amounts:
    total = total + amount
    if amount >= 5000:
        big_orders = big_orders + 1
print("Total:", round(total, 2))
print("Big orders:", big_orders)
```

## 3. Building a new list inside a loop

```python
sizes = []
for amount in order_amounts:
    if amount >= 5000:
        sizes.append("Big")
    elif amount >= 1000:
        sizes.append("Medium")
    else:
        sizes.append("Small")
print(sizes)
```

## 4. Loop over a mini table: cleaning messy city names
These are real spellings from `orders_raw.csv`. A dictionary maps old names to the standard ones.

```python
raw_cities = ["delhi", "New Delhi", "Delhi ", "DELHI", "Bangalore", "Bombay", " Mumbai", "Madras", "Calcutta", "pune"]
rename = {"New Delhi": "Delhi", "Bangalore": "Bengaluru", "Bombay": "Mumbai",
          "Madras": "Chennai", "Calcutta": "Kolkata"}

clean_cities = []
for c in raw_cities:
    c = c.strip().title()          # " Mumbai" -> "Mumbai", "DELHI" -> "Delhi"
    c = rename.get(c, c)           # swap old name if it is in the dictionary, else keep it
    clean_cities.append(c)
print(clean_cities)
```

## 5. Looping over a list of dictionaries: return rate

```python
orders = [
    {"order_id": "O10001", "city": "Delhi",   "returned": "N"},
    {"order_id": "O10002", "city": "Chennai", "returned": "N"},
    {"order_id": "O10003", "city": "Delhi",   "returned": "N"},
    {"order_id": "O10004", "city": "Lucknow", "returned": "Y"},
    {"order_id": "O10005", "city": "Chennai", "returned": "N"},
]
returns = 0
for order in orders:
    if order["returned"] == "Y":
        returns += 1                 # += 1 is short for returns = returns + 1
print(f"Return rate: {returns / len(orders):.0%}")
```

## 6. range() and while (quick look)

```python
for i in range(3):          # 0, 1, 2
    print("Order number", i + 1)

stock = 5                   # 5 Smartphones in the warehouse
while stock >= 2:          # enough for one more order?
    stock = stock - 2       # each order takes 2
    print("Stock left:", stock)
```
