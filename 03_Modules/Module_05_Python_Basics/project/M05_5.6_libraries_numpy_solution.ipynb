## 2. The problem with plain lists
Unit prices and quantities of the first 8 ShopKart orders. To get price x quantity with lists we need a loop.

```python
prices_list = [362, 300, 2258, 2020, 1937, 1440, 15303, 14841]
qty_list = [6, 2, 1, 2, 2, 2, 2, 1]

totals = []
for p, q in zip(prices_list, qty_list):   # zip pairs items up
    totals.append(p * q)
print(totals)

print(prices_list * 2)   # this does NOT double the prices - it repeats the list!
```

## 3. NumPy arrays: maths on every value at once (vectorised)

```python
prices = np.array([362, 300, 2258, 2020, 1937, 1440, 15303, 14841])
qty = np.array([6, 2, 1, 2, 2, 2, 2, 1])
discount = np.array([0.20, 0.05, 0.05, 0.15, 0.10, 0.05, 0.20, 0.05])

gross = prices * qty                     # element by element, no loop
revenue = gross * (1 - discount)
print(gross)
print(revenue.round(2))
```

```python
print("Total revenue:", revenue.sum().round(2))
print("Average order:", revenue.mean().round(2))
print("Biggest order:", revenue.max().round(2))
print("Median order:", np.median(revenue).round(2))
```

```python
# Boolean filtering: which orders are 'Big' (>= 5000)?
print(revenue >= 5000)
print(revenue[revenue >= 5000].round(2))
print("Count of big orders:", (revenue >= 5000).sum())
```

```python
# 18% GST added to every price in one line
(prices * 1.18).round(2)
```

## 4. pandas: NumPy arrays with column names = a table
pandas is built on NumPy. Every column of a pandas table is basically a NumPy array with a name. Let us load the real file.

```python
orders = pd.read_csv("orders_raw.csv")
print(orders.shape)        # (rows, columns)
orders.head()
```

```python
# Each column is like a NumPy array - so maths works on all rows at once
print("Average delivery days:", orders["delivery_days"].mean().round(2))
print("Largest quantity:", orders["quantity"].max())   # 99 items in one order? Suspicious!
```

Notice: we cannot yet do `orders["unit_price"] * orders["quantity"]` reliably because some prices are text like `"₹1,299"`. That, and the suspicious quantity of 99, are the first jobs of Module 6.
