## Exercise 1 - Revenue of one order
Order O10009: 2 Tablets at ₹18,548 each with a 5% discount. Store quantity, unit_price and discount in variables and print the revenue rounded to 2 decimals.

```python
quantity = 2
unit_price = 18548
discount = 0.05
revenue = round(quantity * unit_price * (1 - discount), 2)
print(revenue)
```

## Exercise 7 - if/elif/else
Write code that prints a delivery label for `delivery_days = 6`: 2-3 days is "Fast", 4-5 is "Normal", 6 or more is "Slow".

```python
delivery_days = 6
if delivery_days <= 3:
    label = "Fast"
elif delivery_days <= 5:
    label = "Normal"
else:
    label = "Slow"
print(label)
```

## Exercise 8 - Loop and count
`payments = ["COD", "Debit Card", "Debit Card", "COD", "UPI", "UPI", "UPI", "Credit Card", "Debit Card", "Credit Card"]`. Use a loop to count how many are UPI.

```python
payments = ["COD", "Debit Card", "Debit Card", "COD", "UPI", "UPI", "UPI", "Credit Card", "Debit Card", "Credit Card"]
upi = 0
for p in payments:
    if p == "UPI":
        upi += 1
print(upi)
```

## Exercise 9 - Function
Write `profit(quantity, unit_price, discount, cost_price)` that returns revenue minus quantity x cost_price. Test it on order O10001: 6 x Wings of Fire at ₹362, 20% discount, cost price ₹240.

```python
def profit(quantity, unit_price, discount, cost_price):
    revenue = round(quantity * unit_price * (1 - discount), 2)
    return round(revenue - quantity * cost_price, 2)

print(profit(6, 362, 0.2, 240))
```

## Exercise 10 - NumPy
Make a NumPy array of these revenues: 1737.6, 570.0, 2145.1, 3434.0, 3486.6, 2736.0. Print the total and the revenues above ₹2,500.

```python
import numpy as np
rev = np.array([1737.6, 570.0, 2145.1, 3434.0, 3486.6, 2736.0])
print(rev.sum().round(2))
print(rev[rev > 2500])
```
